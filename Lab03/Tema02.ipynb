Ex. 1 Se aruncă două zaruri, iar rezultatele sunt notate cu X şi Y.
1. Calculaţi probabilitatea ca suma să fie cel puţin 10, ştiind că primul zar indică un număr par.
2. Calculaţi probabilitatea ca suma să fie cel puţin 10, ştiind că cel puţin unul dintre zaruri indică un număr par.
3. Calculaţi probabilitatea ca ambele zaruri să indice numere pare, ştiind că suma lor este 8.
4. Sunt evenimentele A = {X + Y ≥ 10} şi B = {X este par} independente?

Extensie Python: Simulaţi 100.000 de aruncări şi estimaţi probabilităţile condiţionate. Comparaţi rezultatele cu val-
orile exacte.

1.
variabila X = numarul indicat de primul zar \
variabila Y = numarul indicat de al doilea zar \
evenimentul PX = primul zar indica par \
evenimentul M10 = suma este cel putin 10 (**M**acar **10**) \

Formula probabilitatii totale, cu $P(X = k | PX) = 1/3$ pentru fiecare $k$ par:
$$\begin{aligned}
P(M10 | PX) &= P(M10 | X = 2) \cdot \tfrac{1}{3} + P(M10 | X = 4) \cdot \tfrac{1}{3} + P(M10 | X = 6) \cdot \tfrac{1}{3} \\
&= \tfrac{1}{3} \left(P(Y \geq 8) + P(Y \geq 6) + P(Y \geq 4)\right) = \tfrac{1}{3} \left(0 + 1/6 + 3/6\right) = 4/18 = 2/9
\end{aligned}$$

2.
evenimentul PY = al doilea zar indica par \

$P(M10 \cap PX) = P(M10 | PX) \cdot P(PX) = 2/9 \cdot 1/2 = 4/36$, la fel $P(M10 \cap PY) = 4/36$ (simetrie) 

$M10 \cap PX \cap PY = \{(4, 6), (6, 4), (6, 6)\}$.
$$\begin{aligned}
P(M10 | PX \cup PY) &= \frac{P(M10 \cap (PX \cup PY))}{P(PX \cup PY)} = \frac{P(M10 \cap PX) + P(M10 \cap PY) - P(M10 \cap PX \cap PY)}{1 - P(\overline{PX} \cap \overline{PY})} \\
&= \frac{4/36 + 4/36 - 3/36}{1 - 9/36} = \frac{5/36}{27/36} = 5/27
\end{aligned}$$

3.
variabila S = X + Y
vrem sa aflam $P((PX \cap PY) | S = 8)$
S = 8 | (PX \cap PY) = {(2, 6), (4, 4), (6, 2)}
$$
\begin{aligned}
P((PX \cap PY) | S = 8) &= \dfrac{P((PX \cap PY)) * P(S = 8 | (PX \cap PY))}{P(S = 8)} \\
                        &= \dfrac{9/36 \cdot 3/9}{5/36} = \dfrac{3/36}{5/36} = 3/5
\end{aligned}
$$

4.
A = {X + Y $\geq$ 10} = {(4, 6), (5, 5), (5, 6), (6, 4), (6, 5), (6, 6)} $\rightarrow$ P(A) = 6/36 \
B = {X = par} = PX $\rightarrow$ P(B) = 1/2

A si B sunt independente $\iff P(A) * P(B) = P(A \cap B)$ \
$P(A \cap B) = P(A | B) * P(B) = 2 / 9 * 1 / 2 = 2 / 18 = 4 / 36$ \
$P(A) * P(B) = 3 / 36$ 

$P(A) * P(B) \neq P(A \cap B) \rightarrow$ evenimentele nu sunt independente

```python
# Extensie Python
import numpy
from fractions import Fraction

iters = 100000
zaruri = numpy.array([(numpy.random.randint(1, 7), numpy.random.randint(1, 7)) for i in range(iters)])
X = zaruri[:, 0]
Y = zaruri[:, 1]
PX           = len(zaruri[X % 2 == 0]) / iters
PY           = len(zaruri[Y % 2 == 0]) / iters
PXY          = len(zaruri[(X % 2 == 0) & (Y % 2 == 0)]) / iters
PX_sau_Y     = len(zaruri[(X % 2 == 0) | (Y % 2 == 0)]) / iters
M10          = len(zaruri[X + Y >= 10]) / iters
S8           = len(zaruri[X + Y == 8]) / iters
PXY_S8       = len(zaruri[(X + Y == 8) & (X % 2 == 0) & (Y % 2 == 0)]) / iters

M10_PX       = len(zaruri[(X + Y >= 10) & (X % 2 == 0)]) / iters
M10_PX_sau_Y = len(zaruri[(X + Y >= 10) & ((X % 2 == 0) | (Y % 2 == 0))]) / iters

GREEN = '\033[32m'
RESET = '\033[0m'

def afiseaza(nume, simulat, ideal):
    print(f'{nume:<18} = {simulat:.6f}   {GREEN}{float(ideal):.6f} ({ideal}){RESET}')

afiseaza('P(PX)', PX, Fraction(1, 2))
afiseaza('P(PY)', PY, Fraction(1, 2))
afiseaza('P(PXY)', PXY, Fraction(1, 4))
afiseaza('P(M10)', M10, Fraction(1, 6))
afiseaza('P(S8)', S8, Fraction(5, 36))
print()
# 1.
afiseaza('P(M10 | PX)', M10_PX / PX, Fraction(2, 9))
# 2.
afiseaza('P(M10 | (PX U PY))', M10_PX_sau_Y / PX_sau_Y, Fraction(5, 27))
# 3.
afiseaza('P(PXY | S8)', PXY_S8 / S8, Fraction(3, 5))
# 4. A = M10, B = PX
afiseaza('P(M10 ∩ PX)', M10_PX, Fraction(1, 9))
afiseaza('P(M10) * P(PX)', M10 * PX, Fraction(1, 12))
```

$$

P(spam)             &= 0.3 \\

\begin{aligned}
P(oferta | spam)      &= 0.8 \\
P(oferta | \neg spam) &= 0.15
\end{aligned}
\right\} &\Rightarrow
\begin{aligned}

          &= 0.8 * 0.3 + 0.15 * 0.7 = 0.345
\end{aligned} \\
P(spam | oferta)    &= \dfrac{P(oferta | spam) * P(spam)}{P(oferta)} = \dfrac{0.8 * 0.3}{0.345} = 0.69565217391

$$

Ex. 3 Paradoxul lui Simpson
O universitate analizează rata de promovare la două discipline, A şi B, pentru studenţii din două programe de studiu:

$$
\begin{array}{l c c}
\text{Program}     & \text{Disciplina A}   & \text{Disciplina B}   \\
\text{Programul 1} & 81 \text{ din } 90    & 19 \text{ din } 20    \\
\text{Programul 2} & 1 \text{ din } 10     & 16 \text{ din } 80    \\
\end{array}
$$

Valorile reprezintă numărul studenţilor promovaţi din totalul celor evaluaţi.
1. Calculaţi rata de promovare la fiecare disciplină, separat pentru fiecare program.
2. Care disciplină are o rată de promovare mai mare în fiecare program?
3. Calculaţi rata globală de promovare pentru cele două discipline.
4. Explicaţi de ce ordinea disciplinelor se inversează atunci când agregăm datele.
5. Formulaţi fenomenul în termeni de probabilităţi condiţionate.
Extensie Python: Construiţi un exemplu propriu de paradox Simpson şi utilizaţi librăria pandas pentru a analiza
datele.
Notă: Pentru calculele teoretice, încărcaţi calculul fie în Markdown/Latex, fie ca foto la foia de lucru

1.
$$
\begin{aligned}
P(Promovare | (Discp_A \cap Prog_1)) &= \dfrac{81}{90} = 0.9 \\
P(Promovare | (Discp_B \cap Prog_1)) &= \dfrac{19}{20} = 0.95 \\
P(Promovare | (Discp_A \cap Prog_2)) &= \dfrac{1}{10} = 0.1 \\
P(Promovare | (Discp_B \cap Prog_2)) &= \dfrac{16}{80} = 0.2
\end{aligned}
$$

2. In ambele programe, disciplina B are o rata mai mare de promovare

3.
$$
\begin{aligned}
P(Promovare | Discp_A) &= \dfrac{81 + 1}{90 + 10}  = 0.82 \\
P(Promovare | Discp_B) &= \dfrac{19 + 16}{20 + 80} = 0.35 \\
\end{aligned}
$$


4.

Global, Disciplina A are rata de promovare cu mult mai mare.

Cea ce ar face Disciplina A sa para mai usoara, in ciuda faptului ca daca ne uitam per program disciplina B pare mai usoara in ambele cazuri.

Aceasta discrepanta apare din cauza faptului ca nu evaluam destul de specific.
Studentii din programul 2 sunt evident mult mai slab pregatiti decat cei din programul 1, si de asemenea acestia sunt mult mai multi la disciplina B, tragand mult in jos rata de promovare cand calculam agregat.

5.
Pentru fiecare program in parte, disciplina B are rata de promovare mai mare:
$$
P(Promovare | (Discp_A \cap Prog_i)) < P(Promovare | (Discp_B \cap Prog_i)), \quad i \in \{1, 2\}
$$
dar la nivel global ordinea se inverseaza:
$$
P(Promovare | Discp_A) > P(Promovare | Discp_B)
$$

$$
\begin{aligned}
P(Prog_1 | Discp_A) &= \dfrac{90}{100} = 0.9, & P(Prog_2 | Discp_A) &= \dfrac{10}{100} = 0.1 \\
P(Prog_1 | Discp_B) &= \dfrac{20}{100} = 0.2, & P(Prog_2 | Discp_B) &= \dfrac{80}{100} = 0.8
\end{aligned}
$$
$$
\begin{aligned}
P(Promovare | Discp_A) &= \sum_{i=1}^{2} P(Promovare | (Discp_A \cap Prog_i)) \cdot P(Prog_i | Discp_A) = 0.9 \cdot 0.9 + 0.1 \cdot 0.1 = 0.82 \\
P(Promovare | Discp_B) &= \sum_{i=1}^{2} P(Promovare | (Discp_B \cap Prog_i)) \cdot P(Prog_i | Discp_B) = 0.95 \cdot 0.2 + 0.2 \cdot 0.8 = 0.35
\end{aligned}
$$
