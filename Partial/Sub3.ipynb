```python
#a
import numpy as np
import arviz as az
import matplotlib.pyplot as plt
from scipy import stats

alpha_prior, beta_prior = 2, 8
k, n = 5, 20

alpha_post = alpha_prior + k
beta_post  = beta_prior + n - k

prior = stats.beta(a = alpha_prior, b = beta_prior)
posterior = stats.beta(a = alpha_post, b = beta_post)


x = np.linspace(0, 1, 1000)


plt.plot(x, prior.pdf(x), label = f"A priori Beta({alpha_prior}, {beta_prior})")
# plt.axvline(total_calls / n, color = "red", linestyle = ":", label = "Rata observata (18)")
plt.title("Distributia a priori")
plt.legend()
plt.show()

plt.plot(x, posterior.pdf(x), label = f"A posteriori Beta({alpha_post}, {beta_post})")
# plt.axvline(total_calls / n, color = "red", linestyle = ":", label = "Rata observata (18)")
plt.title("Distributia a posteriori")
plt.legend()
plt.show()
```
