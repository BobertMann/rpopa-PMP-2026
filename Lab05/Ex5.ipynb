Un profesor dă teste care pot fi dificile, medii sau uşoare. Probabilitatea dificultăţii primului test este aceeaşi. Dacă, la
un moment dat, dă un test dificil, următorul test poate fi doar mediu sau uşor, cu aceeaşi probabilitate. 
Însă dacă dă un test mediu sau uşor, atunci următorul test va fi dificil cu probabilitate 0.5, sau mediu sau uşor cu aceeaşi probabilitate,
0.25. \
Nota unui student la test, FB, B, S sau NS depinde de dificultatea testului. Astfel, probabilităţile condiţionate ale notei
obţinute, dată fiind dificultatea testului, sunt: 

● test dificil: [0.1, 0.2, 0.4, 0.3];  
● test mediu: [0.15, 0.25, 0.5, 0.1];  
● test uşor: [0.2, 0.3, 0.4, 0.1]. 


a) Utilizaţi librăria hmmlearn pentru a defini modelul Markov ascuns şi desenaţi diagrama de stare.

b) Să presupunem că aţi observat următoarea secvenţă de note: FB, FB, S, B, B, S, B, B, NS, B, B, S. Determinaţi cea
mai probabilă secvenţă de dificultăţi pentru testele corespunzaătoare şi probabilitatea acesteia.

```python
import numpy as np
from hmmlearn import hmm
np.random.seed(42)

model = hmm.CategoricalHMM(n_components = 3, n_iter = 12)

states    = ['Test usor', 'Test mediu', 'Test dificil']
emissions = ['FB', 'B', 'S', 'NS'] 

model.startprob_ = np.array([1/3, 1/3, 1/3], dtype = np.float64)

model.transmat_= np.array([[0.25, 0.25, 0.5],
                           [0.25, 0.25, 0.5],
                           [0.5, 0.5, 0]], dtype = np.float64)

model.emissionprob_ = np.array([[0.2, 0.3, 0.4, 0.1],
                       [0.15, 0.25, 0.5, 0.1],
                       [0.1, 0.2, 0.4, 0.3]], dtype = np.float64)


observations = np.array([0, 0, 2, 1, 1, 2, 1, 1, 3, 1, 1, 2]).reshape(-1, 1)


log_prob, predicted_states = model.decode(observations)
decoded_predicted = [states[i] for i in predicted_states]

print("\nPredicted Hidden States (Viterbi):")
print(decoded_predicted)

print("\nProbability of the most likely sequence:")
print(f"log P = {log_prob:.4f}")
print(f"P = {np.exp(log_prob):.6e}")
```
