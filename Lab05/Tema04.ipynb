## Exercițiul 1. De la independență condiționată la factorizare

Considerăm trei variabile aleatoare discrete $A, B, C$ și graful neorientat $A - B - C$. Presupunem că distribuția comună este **strict pozitivă** pentru toate configurațiile posibile.



a) Explicați, în cuvintele voastre, ce înseamnă $A \perp C \mid B$. Ce informație suplimentară despre $A$ mai oferă $C$ atunci când cunoaștem $B$?



$A \perp C \mid B$ inseamnă că $A$ este independent de $C$ atunci când îl cunoaștem pe B. Înseamnă că atunci când îl cunoaștem pe B, C nu mai oferă deloc informație despre A.

În cuvintele mele, $A \perp C$ înseamnă că cele două variabile pot trăi fără a ști absolut nimic una despre alta, și nu putem afla nici o informație despre una prin a o afla pe cealaltă. Independența condiționată, descrie același concept, dar pe un spațiu (sau spații) mai restrâns de valori, anume cel determinat de B (câte unul pentru fiecare valoare a lui B). 

b) Pornind de la formula probabilității condiționate, demonstrați că $A \perp C \mid B$ implică existența unor potențiale strict pozitive $\phi_{AB}$ și $\phi_{BC}$ și a unei constante $Z > 0$ pentru care

$$P(a, b, c) = \frac{1}{Z}\,\phi_{AB}(a, b)\,\phi_{BC}(b, c).$$

*Indiciu:* folosiți $P(a, b, c) = P(a \mid b)\,P(c \mid b)\,P(b)$.



Distribuția comună este strict pozitivă $\Rightarrow$ $P(b) > 0$ pentru orice $b$, deci probabilitățile condiționate în raport cu $B$ sunt bine definite. Din definiția probabilității condiționate:

$$P(a, b, c) = P(a, c \mid b)\,P(b).$$

Din $A \perp C \mid B$ avem $P(a, c \mid b) = P(a \mid b)\,P(c \mid b)$, deci

$$P(a, b, c) = P(a \mid b)\,P(c \mid b)\,P(b).$$

Scriem $P(b) = \sqrt{P(b)} \cdot \sqrt{P(b)}$ și grupăm factorii:

$$P(a, b, c) = \underbrace{P(a \mid b)\sqrt{P(b)}}_{\phi_{AB}(a, b)} \; \underbrace{P(c \mid b)\sqrt{P(b)}}_{\phi_{BC}(b, c)}.$$

Alegem:

$$\phi_{AB}(a, b) = P(a \mid b)\sqrt{P(b)}, \qquad \phi_{BC}(b, c) = P(c \mid b)\sqrt{P(b)}, \qquad Z = 1.$$


- $\phi_{AB}$ depinde doar de $a$ și $b$, iar $\phi_{BC}$ doar de $b$ și $c$, deci fiecare potențial corespunde unei muchii din graful $A - B - C$.
- Potențialele sunt strict pozitive: $P(a \mid b) = \frac{P(a, b)}{P(b)}$, unde $P(a, b) = \sum_c P(a, b, c) > 0$, iar $\sqrt{P(b)} > 0$. Analog pentru $P(c \mid b)$.
- $Z = 1 > 0$ și este chiar constanta de normalizare: $\sum_{a, b, c} \phi_{AB}(a, b)\,\phi_{BC}(b, c) = \sum_{a, b, c} P(a, b, c) = 1$.

Prin urmare $P(a, b, c) = \frac{1}{Z}\,\phi_{AB}(a, b)\,\phi_{BC}(b, c)$, cu $Z = 1$.

c) Demonstrați și reciproca, pornind de la factorizarea de mai sus. *Indiciu:* calculați $P(a, c \mid b)$ și verificați dacă este produsul lui $P(a \mid b)$ și $P(c \mid b)$.

**Predare:** o demonstrație scurtă, în LaTeX/Markdown sau ca fotografie lizibilă. Nu este necesar cod Python.

Pornim de la factorizarea

$$P(a, b, c) = \frac{1}{Z}\,\phi_{AB}(a, b)\,\phi_{BC}(b, c), \qquad \phi_{AB}, \phi_{BC} > 0, \quad Z > 0.$$

Notăm, pentru fiecare valoare $b$ a lui $B$:

$$S_{AB}(b) = \sum_{a \in Val(A)} \phi_{AB}(a, b), \qquad S_{BC}(b) = \sum_{c \in Val(C)} \phi_{BC}(b, c).$$

Potențialele sunt strict pozitive $\Rightarrow$ $S_{AB}(b) > 0$ și $S_{BC}(b) > 0$.

Distribuțiile marginale se obțin sumând după variabilele lipsă. Cum $\phi_{AB}(a, b)$ nu depinde de $c$, iar $\phi_{BC}(b, c)$ nu depinde de $a$, putem scoate fiecare potențial în afara sumei după variabila de care nu depinde:

$$P(a, b) = \sum_{c} P(a, b, c) = \frac{1}{Z}\,\phi_{AB}(a, b) \sum_{c} \phi_{BC}(b, c) = \frac{1}{Z}\,\phi_{AB}(a, b)\,S_{BC}(b),$$

$$P(b, c) = \sum_{a} P(a, b, c) = \frac{1}{Z}\,\phi_{BC}(b, c) \sum_{a} \phi_{AB}(a, b) = \frac{1}{Z}\,S_{AB}(b)\,\phi_{BC}(b, c),$$

$$P(b) = \sum_{a} \sum_{c} P(a, b, c) = \frac{1}{Z} \left(\sum_{a} \phi_{AB}(a, b)\right) \left(\sum_{c} \phi_{BC}(b, c)\right) = \frac{1}{Z}\,S_{AB}(b)\,S_{BC}(b) > 0.$$

De aici, probabilitățile condiționate (constanta $Z$ se simplifică):

$$P(a \mid b) = \frac{P(a, b)}{P(b)} = \frac{\phi_{AB}(a, b)\,S_{BC}(b)}{S_{AB}(b)\,S_{BC}(b)} = \frac{\phi_{AB}(a, b)}{S_{AB}(b)},$$

$$P(c \mid b) = \frac{P(b, c)}{P(b)} = \frac{S_{AB}(b)\,\phi_{BC}(b, c)}{S_{AB}(b)\,S_{BC}(b)} = \frac{\phi_{BC}(b, c)}{S_{BC}(b)}.$$

Calculăm acum $P(a, c \mid b)$:

$$P(a, c \mid b) = \frac{P(a, b, c)}{P(b)} = \frac{\frac{1}{Z}\,\phi_{AB}(a, b)\,\phi_{BC}(b, c)}{\frac{1}{Z}\,S_{AB}(b)\,S_{BC}(b)} = \frac{\phi_{AB}(a, b)}{S_{AB}(b)} \cdot \frac{\phi_{BC}(b, c)}{S_{BC}(b)} = P(a \mid b)\,P(c \mid b).$$

$\Rightarrow$  $A \perp C \mid B$.

## Exercițiul 3. Restaurarea unei imagini binare cu un MRF

Dorim să reconstruim o imagine binară mică, afectată de zgomot. Fiecare pixel are valoarea 0 (negru) sau 1 (alb). Vom utiliza o rețea Markov pe o grilă $5 \times 5$, în care fiecare pixel este conectat numai cu vecinii săi direcți (sus, jos, stânga, dreapta).

c) Modelați pixelii necunoscuți prin variabile binare $x_i$. Construiți graful MRF și definiți energia

$$E(x; y) = \lambda \sum_i (x_i - y_i)^2 + \sum_{(i,j) \in E} (x_i - x_j)^2, \qquad \lambda > 0.$$

Explicați rolul fiecărui termen și scrieți potențialele unare $\phi_i(x_i) = e^{-\lambda (x_i - y_i)^2}$ și pereche $\phi_{ij}(x_i, x_j) = e^{-(x_i - x_j)^2}$.

Fiecărui pixel $i$ îi asociem o variabilă binară $x_i \in \{0, 1\}$ (valoarea reconstruită), iar $y_i$ este valoarea observată, cu zgomot. Graful MRF are 25 de noduri, câte unul pentru fiecare pixel, și o muchie între oricare doi pixeli vecini pe orizontală sau pe verticală: $5 \cdot 4 = 20$ muchii orizontale și $20$ verticale, deci $|E| = 40$.

Rolul termenilor din energie:
- $\lambda \sum_i (x_i - y_i)^2$. Cum $(x_i - y_i)^2 = 1$ dacă $x_i \ne y_i$ și $0$ dacă $x_i = y_i$ termenul numără pixelii schimbați față imaginea dată, fiecare având "cost" $\lambda$. -> ține imaginea reconstruită aproape de cea observată.
- $\sum_{(i,j) \in E} (x_i - x_j)^2$ numără perechile de vecini cu valori diferite, adică penalizează pixelii vecini care diferă (proprietatea de care ne folosim pentru a reconstrui imaginea este că pixelii vecini tind sa aibă aceași culoare).
- $\lambda$ factor care ne permite modularea costului de reconstituire al unui pixel: $\lambda$ mare $\Rightarrow$ $x$ rămâne aproape de $y$, $\lambda$ mic $\Rightarrow$ netezim mai tare.

Cum $P(x \mid y) \propto e^{-E(x; y)}$, iar exponențiala unei sume este produsul exponențialelor,

$$P(x \mid y) \propto \prod_i \phi_i(x_i) \prod_{(i,j) \in E} \phi_{ij}(x_i, x_j),$$

cu potențialele

$$\phi_i(x_i) = e^{-\lambda (x_i - y_i)^2} = \begin{cases} 1, & x_i = y_i \\ e^{-\lambda}, & x_i \ne y_i \end{cases}, \qquad \phi_{ij}(x_i, x_j) = e^{-(x_i - x_j)^2} = \begin{cases} 1, & x_i = x_j \\ e^{-1}, & x_i \ne x_j \end{cases}.$$

Deci minimizarea energiei $E(x; y)$ este echivalentă cu maximizarea lui $P(x \mid y)$ (MAP).

d) Folosind `pyAgrum` sau `pgmpy`, construiți factorii și determinați o configurație MAP, adică o imagine $x^*$ care minimizează $E(x; y)$. Pentru grila $5 \times 5$, inferența exactă poate fi costisitoare; este permisă și o metodă aproximativă (de exemplu, ICM), dacă explicați algoritmul folosit.

````python

def nume(r, c):
  return f'x{r}{c}'

muchii_grila = [((r, c), (r, c + 1)) for r in range(5) for c in range(4)] + \
               [((r, c), (r + 1, c)) for r in range(4) for c in range(5)]

def energie(x, y, lam):
  fidelitate = lam * np.sum((x - y) ** 2)
  netezire = sum((x[a] - x[b]) ** 2 for a, b in muchii_grila)
  return fidelitate + netezire

def build_mrf(y, lam):
  mrf = pyagrum.MarkovRandomField()
  for r in range(5):
    for c in range(5):
      mrf.add(pyagrum.LabelizedVariable(nume(r, c), nume(r, c), ['0', '1']))

  # factori unari: phi_i(x_i) = exp(-lam * (x_i - y_i)^2), pentru x_i = 0, 1
  for r in range(5):
    for c in range(5):
      mrf.addFactor([nume(r, c)]).fillWith([math.exp(-lam * (x - y[r, c]) ** 2) for x in (0, 1)])

  # factori pereche: phi_ij(x_i, x_j) = exp(-(x_i - x_j)^2), simetric deci ordinea nu conteaza
  for a, b in muchii_grila:
    mrf.addFactor([nume(*a), nume(*b)]).fillWith([1, math.exp(-1), math.exp(-1), 1])

  return mrf

def build_and_print_mrf(y, lam):
  mrf = build_mrf(y, lam)
  print(mrf)

  plt.figure(figsize=(4, 4))
  for a, b in muchii_grila:
    plt.plot([a[1], b[1]], [a[0], b[0]], color='gray', zorder=1)
  for r in range(5):
    for c in range(5):
      plt.scatter(c, r, s=700, c='white' if y[r, c] else 'black', edgecolors='gray', zorder=2)
      plt.text(c, r, nume(r, c), ha='center', va='center', fontsize=8,
              color='black' if y[r, c] else 'white', zorder=3)
  plt.xlim(-0.6, 4.6)
  plt.ylim(4.6, -0.6)
  plt.axis('off')
  plt.title('Graful MRF (culoare = y)')
  plt.show()

  return mrf
build_and_print_mrf(y, 1)
pass
````

Folosim inferență exactă. `ShaferShenoyMRFInference` nu are o metodă directă pentru MAP (doar `posterior` și `jointPosterior`), așa că cerem distribuția comună a tuturor celor 25 de pixeli, $P(x \mid y)$, și luăm configurația cu probabilitatea maximă: $x^* = \arg\max_x P(x \mid y) = \arg\min_x E(x; y)$.

Distribuția comună are $2^{25} \approx 33.5$ milioane de valori, deci pe grila $5 \times 5$ durează câteva secunde și ~1 GB de memorie. Pe o grilă mai mare numărul de valori crește exponențial și nu ar mai fi fezabil; de aceea se folosesc metode aproximative precum ICM.

````python
def map_mrf(y, lam):
  mrf = build_mrf(y, lam)
  pixeli = {nume(r, c) for r in range(5) for c in range(5)}

  ie = pyagrum.ShaferShenoyMRFInference(mrf)
  ie.addJointTarget(pixeli)  # distributia comuna a tuturor pixelilor
  ie.makeInference()

  maxime, p_max = ie.jointPosterior(pixeli).argmax()
  x = np.array([[maxime[0][nume(r, c)] for c in range(5)] for r in range(5)])
  return x, p_max

def imc_mrf(y, lam):
  mrf = build_mrf(y, lam)

  reconstruire = y.copy()

  while True:
    ok = False
    for r in range(5):
      for c in range(5):
        prev_energy = energie(reconstruire, y, lam)
        reconstruire[r][c] = 1 - reconstruire[r][c]
        if energie(reconstruire, y, lam) < prev_energy :
          ok = True
        else:
          reconstruire[r][c] = 1 - reconstruire[r][c]
    if ok == False :
      break
  return reconstruire

x_map, p_max = map_mrf(y, 1)
print(x_map)
print(f'P(x* | y) = {p_max:.4f}, E(x*; y) = {energie(x_map, y, 1)}')

fig, axes = plt.subplots(1, 3, figsize=(9, 3))
for ax, img, titlu in zip(axes, [original, y, x_map], ['Original', 'Observat (y)', f'MAP (λ = {1})']):
  ax.imshow(img, cmap='gray', vmin=0, vmax=1)
  ax.set_title(titlu)
  ax.axis('off')
plt.show()

````
